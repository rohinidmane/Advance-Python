```python
Pandas:Pandas is open-source Python library which is used for data manipulation and analysis. 
It consist of data structures and functions to perform efficient operations on data. 
It is well-suited for working with tabular data such as spreadsheets or SQL tables.
```

```python
# Importing Pandas
import pandas as pd
import numpy as np
```

```python
Data Structures in Pandas Library
Pandas provide two data structures for manipulating data which are as follows:
```

```python
1. Pandas Series
A Pandas Series is one-dimensional labeled array capable of holding data of any type (integer, string, float, Python objects etc.). 
The axis labels are collectively called indexes.
```

```python
#Create first pandas series

number_series = pd.Series([2, 3, 4, 5])
print(number_series)
```

```python
number_series = pd.Series([2, 3, 4, 5], index = ['a', 'b', 'c', 'd'], name = 'Number')
print(number_series)
```

```python
#accessing pandas series elements
print(number_series['c'])
print(number_series.iloc[1])
```

```python
2. Pandas DataFrame:Pandas DataFrame is a two-dimensional data structure with labeled axes (rows and columns)
```

```python
#Creating our first DataFrame using 2-D Numpy Array
```

```python
twod_array = np.random.randint(2, 10,(3,4))
twod_array
```

```python
twod_array = np.random.randint(2, 10,(3,4))
first_dataframe = pd.DataFrame(twod_array, columns =['Akshay', 'Devashish', 'Divy', 'Monika'], index =['a','b','c'])
first_dataframe
```

```python
# Initialise list of lists
list_of_lists = [['Ashu', 10], ['Raghav',20], ['Shailja',30]]

# create the pandas dataframe
df = pd.DataFrame(list_of_lists, columns=['Name','Age'])

# print dataframe
df
```

```python
# creating dataframe using Dictionary of lists(column wise) --- Dictionary of Lists must have equal-length lists:
import pandas as pd

# store the data in the dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Paris']
}

# create dataframe
df = pd.DataFrame(data)

#print datframe
print(df)
```
