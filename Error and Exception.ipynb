```python
# Python code to illustrate
# working of try()
def divide(x, y):
    try:
        # Floor Division : Gives only Fractional Part as Answer
        result = x // y
        print("Yeah ! Your answer is :", result)
    except ZeroDivisionError:
        print("Sorry ! You are dividing by zero ")

divide(17, 2)
divide(6,0)
```

```python
Else Clause
```

```python
# Program to depict else clause with try-except
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Division by zero error!")
else:
    print(f"Division successful, result is {result}")
```

```python
my_list = [10, 20, 30]

try:
    index = int(input("Enter the index you want to access (0-2): "))
    value = my_list[index]
except IndexError:
    print("Index out of range! Please enter a valid index.")
except ValueError:
    print("Invalid input! Please enter a number.")
else:
    print(f"Element at index {index} is {value}")
```

```python
Using Finally
```

```python
# Python program to demonstrate finally

# No Exception raised in try block
try:
    k = 5//0 # raises divide by zero exception.
    print(k)

# handles zerodivision exception
except ZeroDivisionError:
    print("Can't divide by zero")

finally:
    # this block is always executed
    # regardless of exception generation.
    print('This is always executed')
```

```python
nested try block -- try inside try block
```

```python
try:
    print("Outer try block")
    num = int(input("Enter a number: "))

    try:
        result = 10 / num  # Might raise ZeroDivisionError
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Inner except: Cannot divide by zero")

except ValueError:
    print("Outer except: Invalid input! Not a number.")
```

```python
try:

    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))

    try:
        result = num1 / num2
    except ZeroDivisionError:
        print("Inner error: Cannot divide by zero.")
    else:
        print(f"Inner success: Result is {result}")
    finally:
        print("Inner finally: Division attempt finished.")

except ValueError:
    print("Outer error: Please enter valid integers.")
else:  #else in both blocks runs only if no error occurred.finally in both blocks runs always.
    print("Outer success: Both inputs are valid integers.")
finally:
    print("Outer finally: Program execution completed.")
```
