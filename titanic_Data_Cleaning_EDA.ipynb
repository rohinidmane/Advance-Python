```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset('titanic')
df
```

```python
#Dataset Overview


print('Shape:', df.shape)
df.info()
df.isnull().sum()
```

```python
#Handling Missing Values

# Drop missing rows
df_drop = df.dropna()
print(df.shape, df_drop.shape)

# Fill missing values
df['age_mean'] = df['age'].fillna(df['age'].mean())
df['age_median'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
#df['deck'] = df['deck'].cat.add_categories('Unknown')
#df['deck'] = df['deck'].astype('str').fillna('Unknown')
df['deck'] = df['deck'].astype('str').replace('nan','Unknown')
df[['age','age_mean','age_median','embarked','deck']].head()
```

```python
#Handling Duplicates

df_dup = pd.concat([df, df.iloc[:5]], ignore_index=True)
print('Duplicate rows:', df_dup.duplicated().sum())
df_dup_clean = df_dup.drop_duplicates()
print(df_dup_clean.shape)
```

```python
#Handling Inconsistent Categories

sample = pd.DataFrame({'Gender':['Male','male','M','Female',' female ','F']})
sample['Gender'] = sample['Gender'].str.strip().str.lower()
mapping={'male':'Male','m':'Male','female':'Female','f':'Female'}
sample['Gender']=sample['Gender'].replace(mapping)
sample
```

```python
#Stripping Whitespace

dirty = pd.DataFrame({'Name':[' John ',' Alice','Bob ']})
dirty['Name']=dirty['Name'].str.strip()
dirty
```

```python
#Incorrect Data Formats

sales = pd.DataFrame({
    'Date':['2024-01-01','01/02/2024','March 3 2024'],
    'Price':['100','$250','300']
})

# Better date parsing
sales['Date'] = pd.to_datetime(
    sales['Date'],
    format='mixed'   # handles mixed date styles
)

# Clean price column
sales['Price'] = (
    sales['Price']
      .replace('[$]', '', regex=True)
      .astype(float)
)

print(sales)
```

```python
#Renaming Columns for Consistency

messy = pd.DataFrame({'Customer Name':['A'],'Order Amount':['100'],'Phone Number':['1234']})
messy.columns = messy.columns.str.strip().str.lower().str.replace(' ','_')
messy.columns
```

```python
#Complete Data Cleaning Pipeline

clean_df = sns.load_dataset('titanic')
clean_df=clean_df.drop_duplicates()
clean_df['age']=clean_df['age'].fillna(clean_df['age'].median())
clean_df['embarked']=clean_df['embarked'].fillna(clean_df['embarked'].mode()[0])
clean_df=clean_df.drop(columns=['deck'])
clean_df.columns=clean_df.columns.str.lower().str.strip().str.replace(' ','_')
clean_df.head()
```

```python
#Bonus Reusable Function

def basic_cleaning(df):
    df=df.drop_duplicates()
    for col in df.select_dtypes(include=['str']):
        df[col]=df[col].astype(str).str.strip()
    for col in df.select_dtypes(include='number'):
        df[col]=df[col].fillna(df[col].median())
    df.columns=df.columns.str.lower().str.replace(' ','_')
    return df

basic_cleaning(sns.load_dataset('titanic')).head()
```
