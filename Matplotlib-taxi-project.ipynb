```python
# NYC Taxi Trips Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv')
```

```python
df
```

```python
# -----------------------------
# 1. Basic Overview
# -----------------------------
print("\n--- BASIC INFO ---")
print(df.info())
#print(df.describe())
```

```python
print("\n--- DESCRIBE ---")
print(df.describe(include='all'))
```

```python
# -----------------------------
# 3. Missing Values & Duplicates
# -----------------------------
print("\n--- MISSING VALUES ---")
print(df.isnull().sum())
```

```python
print("\n--- DUPLICATES ---")
print(df.duplicated().sum())
df
```

```python
# -----------------------------
# 4. Data Type Fixing
# -----------------------------

df['pickup'] = pd.to_datetime(df['pickup'], errors='coerce')
df['dropoff'] = pd.to_datetime(df['dropoff'], errors='coerce')

df.info()
```

```python
# -----------------------------
# 5. Feature Engineering
# -----------------------------

# Trip duration in minutes
df['trip_duration'] = (df['dropoff'] - df['pickup']).dt.total_seconds() / 60

# Time features
df['pickup_hour'] = df['pickup'].dt.hour
df['pickup_day'] = df['pickup'].dt.day_name()

df
```

```python
# -----------------------------
# 7. Univariate Analysis
# -----------------------------

# Fare distribution
plt.figure()
sns.histplot(df['fare'], bins=40, kde=True)
plt.title('Fare Distribution')
plt.show()

# Trip duration
plt.figure()
sns.histplot(df['trip_duration'], bins=40, kde=True)
plt.title('Trip Duration Distribution')
plt.show()

# Distance
plt.figure()
sns.histplot(df['distance'], bins=40, kde=True)
plt.title('Distance Distribution')
plt.show()

# Pickup hour
plt.figure()
sns.countplot(x='pickup_hour', data=df)
plt.title('Trips by Hour')
plt.show()
```

```python

# -----------------------------
# 8. Bivariate Analysis
# -----------------------------

# Fare vs duration
plt.figure()
sns.scatterplot(x='trip_duration', y='fare', data=df, alpha=0.3)
plt.title('Fare vs Trip Duration')
plt.show()

# Fare vs distance
plt.figure()
sns.scatterplot(x='distance', y='fare', data=df, alpha=0.3)
plt.title('Fare vs Distance')
plt.show()

# Boxplot: fare by hour
plt.figure()
sns.boxplot(x='pickup_hour', y='fare', data=df)
plt.title('Fare by Pickup Hour')
plt.show()
```

```python
# -----------------------------
# 9. Multivariate Analysis
# -----------------------------

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()

# Pairplot (sample for performance)
sample_df = df.sample(500, random_state=42)
sns.pairplot(sample_df[['fare', 'distance', 'trip_duration']])
plt.show()
```

```python
# -----------------------------
# 10. Time-Based Analysis
# -----------------------------

# Avg fare by hour
hourly_fare = df.groupby('pickup_hour')['fare'].mean()

plt.figure()
sns.lineplot(x=hourly_fare.index, y=hourly_fare.values)
plt.title('Average Fare by Hour')
plt.xlabel('Hour')
plt.ylabel('Fare')
plt.show()

# Trips by day
plt.figure()
sns.countplot(x='pickup_day', data=df, order=df['pickup_day'].value_counts().index)
plt.xticks(rotation=45)
plt.title('Trips by Day')
plt.show()
```

```python
# -----------------------------
# 11. Key Insights
# -----------------------------

print("\n--- KEY INSIGHTS ---")
print("Average Fare:", df['fare'].mean())
print("Average Distance:", df['distance'].mean())
print("Average Trip Duration:", df['trip_duration'].mean())
print("Peak Hour:", df['pickup_hour'].mode()[0])

print("\nEDA COMPLETED SUCCESSFULLY 🚀")
```
