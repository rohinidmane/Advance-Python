```python
import matplotlib.pyplot as plt

#Data
names = ['Aman', 'Pratiksha', 'Zeehan', 'Denim']
salaries = [20000, 15000, 80000, 50000]

# Create a bar chart
plt.bar(names, salaries, color = 'skyblue')


# Add labels and title
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart Example')

plt.grid()

# show the plot
plt.show()
```

```python
# Pie Chart

import matplotlib.pyplot as plt
x = ['Python', 'Java', 'C++', 'JavaScript']
y = [40, 30, 20, 10]
plt.pie(y, labels=x, autopct='%1.2f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen']

# create a pie chart
plt.pie(sizes, labels = labels, colors =  colors, autopct = '%1.1f%%', startangle = 180, counterclock = False)

# Add title
plt.title("Pie Chart Example")

# show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# generate random data
data = np.random.randint(1000, size = 500)  # size is the no. of samples

# create histogram
plt.hist(data, bins=30, color = 'green', edgecolor = 'black')

# Add labels and titles
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram Example')

# show the plot
plt.show()
```

```python
Multiple Plot
```

```python
import matplotlib.pyplot as plt
import numpy as np

# data
x = np.linspace(0,10,100) #The line x = np.linspace(0, 10, 100) creates an array of 100 evenly spaced values between 0 and 10 (inclusive) using NumPy.
y1 = np.sin(x)
y2 = np.cos(x)

# create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (8, 5))

#plot1
ax1.plot(x, y1, color = 'blue')
ax1.set_title('Sine wave')

# plot2
ax2.plot(x, y2, color='red')
ax2.set_title("Cosine wave")

# Add common lables
fig.suptitle("Multiple Plots Example")
plt.show()
```

```python
import matplotlib.pyplot as plt
# data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# create a plot
plt.plot(x, y, marker= '*', linestyle = '--', color = 'b', label = 'Prime Numbers')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("Saving the Basic line Plot")
plt.legend()
plt.grid()

# save the plot to the file
plt.savefig("my_plot.jpg", dpi = 300)

# show the plot
plt.show()
```
